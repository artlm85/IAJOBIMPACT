```python
plt.figure(figsize=(10,6))

genre_counts = df['Industry'].value_counts()

sns.barplot(
    x=genre_counts.values,
    y=genre_counts.index
)

plt.title('Industrias Mais Frequentes')
plt.xlabel('Quantidade')
plt.ylabel('Tipo de Industria')

plt.show()
```

```python
top_10 = df.sort_values('Industry', ascending=False).head(10)

plt.figure(figsize=(10, 7))

# Criar o gráfico de barras
plt.bar(x=top_10['Job_Title'], height=top_10['AI_Replacement_Risk'], color='skyblue')

plt.xlabel('Profissão')
plt.ylabel('Risco de Substituição ')
plt.title('Top 10 ')
plt.xticks(rotation=45, ha='right')
plt.show()
```
